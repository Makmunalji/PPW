# Singular-Value Decomposition

Dekomposisi Nilai-Singular, atau disingkat SVD, adalah metode dekomposisi matriks untuk mereduksi matriks menjadi bagian-bagian penyusunnya agar perhitungan matriks berikutnya menjadi lebih sederhana.

Untuk kasus kesederhanaan kita akan fokus pada SVD untuk matriks bernilai riil dan mengabaikan kasus untuk bilangan kompleks.

A = U . Sigma . V^T

Di mana A merupakan matriks mxn riil yang hendak didekomposisi, U merupakan matriks mxm, Sigma (sering dilambangkan dengan huruf Yunani kapital Sigma) merupakan matriks diagonal mxn, dan V^T merupakan transpos matriks nxn di mana T merupakan superskrip.

Nilai diagonal dalam matriks Sigma dikenal sebagai nilai singular dari matriks asli A. Kolom-kolom matriks U disebut vektor singular kiri dari A, dan kolom-kolom dari V disebut vektor singular kanan dari A.

SVD dihitung melalui metode numerik iteratif. Kami tidak akan membahas detail metode ini. Setiap matriks persegi panjang memiliki dekomposisi nilai singular, meskipun matriks yang dihasilkan mungkin berisi bilangan kompleks dan keterbatasan aritmatika floating point dapat menyebabkan beberapa matriks gagal terurai dengan rapi.

SVD digunakan secara luas baik dalam perhitungan operasi matriks lainnya, seperti invers matriks, tetapi juga sebagai metode reduksi data dalam pembelajaran mesin. SVD juga dapat digunakan dalam regresi linier kuadrat terkecil, kompresi gambar, dan penghilangan derau pada data.

# Hitung Dekomposisi Nilai Singular

SVD dapat dihitung dengan memanggil fungsi svd()

Fungsi ini mengambil matriks dan mengembalikan elemen U, Sigma, dan V^T. Matriks diagonal Sigma dikembalikan sebagai vektor nilai singular. Matriks V dikembalikan dalam bentuk transposisi, misalnya VT

Contoh berikut mendefinisikan matriks 3×2 dan menghitung dekomposisi nilai singular.

```python
# Singular-value decomposition
from numpy import array
from scipy.linalg import svd
# define a matrix
A = array([[1, 2], [3, 4], [5, 6]])
print("matrix awal : ",A)
# SVD
U, s, VT = svd(A)
print("U = :",U)
print("s = :",s)
print("VT = :",VT)
```

Menjalankan contoh tersebut mula-mula akan mencetak matriks 3x2 yang telah ditentukan, kemudian matriks U 3x3, vektor Sigma 2 elemen, dan elemen matriks V^T 2x2 yang dihitung dari dekomposisi.

# Rekonstruksi Matriks dari SVD

Matriks asli dapat direkonstruksi dari elemen U, Sigma, dan V^T.

Elemen U, s, dan V yang dikembalikan dari svd() tidak dapat dikalikan secara langsung.

Vektor s harus diubah menjadi matriks diagonal menggunakan fungsi diag(). Secara default, fungsi ini akan membuat matriks persegi berukuran nxn, relatif terhadap matriks asli kita. Hal ini menyebabkan masalah karena ukuran matriks tidak sesuai dengan aturan perkalian matriks, di mana jumlah kolom dalam matriks harus sesuai dengan jumlah baris dalam matriks berikutnya.

Setelah membuat matriks diagonal Sigma persegi, ukuran matriks tersebut relatif terhadap matriks mxn asli yang sedang kita dekomposisi, sebagai berikut:

U (m x m) . Sigma (n x n) . V^T (n x n)

Di mana, pada kenyataannya, kita memerlukan:

U (m x m) . Sigma (m x n) . V^T (n x n)

Kita dapat memperolehnya dengan membuat matriks Sigma baru dengan semua nilai nol yang berukuran mxn (misalnya lebih banyak baris) dan mengisi bagian nxn pertama dari matriks tersebut dengan matriks diagonal persegi yang dihitung melalui diag().

```python
# Reconstruct SVD
from numpy import array
from numpy import diag
from numpy import dot
from numpy import zeros
from scipy.linalg import svd
# define a matrix
A = array([[1, 2], [3, 4], [5, 6]])
print(A)
# Singular-value decomposition
U, s, VT = svd(A)
# create m x n Sigma matrix
Sigma = zeros((A.shape[0], A.shape[1]))
# populate Sigma with n x n diagonal matrix
Sigma[:A.shape[1], :A.shape[1]] = diag(s)
# reconstruct matrix
B = U.dot(Sigma.dot(VT))
print(B)
```

A adalah sebuah matriks dengan ukuran 3×2, yaitu tiga baris dan dua kolom.
Matriks A berisi elemen-elemen sebagai berikut:
𝐴=[[1, 2], [3, 4], [5, 6]]

svd(A) melakukan dekomposisi nilai singular pada matriks A, menghasilkan tiga komponen:

- U: Matriks vektor singular kiri, berukuran 3×3.
- s: Vektor nilai singular, dengan panjang 2 (karena SVD menghasilkan nilai singular sebanyak dimensi minimum dari matriks asli, yaitu 2 dalam kasus ini).
- VT: Matriks transpos dari vektor singular kanan, berukuran 2×2.

Setelah SVD, kita memiliki hubungan:

                    A≈U⋅Σ⋅VT
                    
Di mana Σ adalah matriks diagonal yang berisi nilai-nilai singular.

- Sigma = zeros((A.shape[0], A.shape[1])) membuat matriks Sigma berukuran 3×2, yang diisi dengan nol.
- Ukuran ini mengikuti bentuk matriks A, dengan tujuan untuk merekonstruksi A dari dekomposisi nilai singular.

- diag(s) menghasilkan matriks diagonal dengan elemen-elemen dari vektor s pada diagonal utamanya.

- Sigma[:A.shape[1], :A.shape[1]] = diag(s) memasukkan nilai-nilai singular dari s pada bagian atas Sigma yang berukuran 2×2, meninggalkan baris ketiga dari Sigma tetap nol.

Hasil akhir Sigma setelah pengisian akan berbentuk:

Σ=[[s1,0],[0,s2],[0,0]]

Di mana 𝑠 dan 𝑠 adalah nilai-nilai singular dari s.

- Sigma.dot(VT) mengalikan Sigma dengan VT, menghasilkan matriks perantara berukuran 3×2.

- U.dot(...) kemudian mengalikan hasil tersebut dengan U, merekonstruksi matriks B yang seharusnya mendekati matriks asli A.

- Matriks B yang dihasilkan akan berukuran 3×2, sama dengan ukuran matriks asli A.

```python
import pandas as pd
from numpy import array, diag, dot, zeros
from scipy.linalg import svd

# Memuat matriks
dfTFIDF = pd.read_csv('VSM dengan Pembobotan.csv')
dfTFIDF = dfTFIDF.drop("Unnamed: 0", axis='columns')
```

```python
# Melakukan Singular Value Decomposition
U, s, VT = svd(dfTFIDF)
```

SVD(dfTFIDF) melakukan dekomposisi nilai singular terhadap dfTFIDF.
- Hasilnya:
    - U adalah matriks berukuran 100×100, yang berisi vektor-vektor singular kiri.
    - s adalah vektor dari nilai-nilai singular, dengan panjang 100 (karena SVD menghasilkan nilai singular sebanyak dimensi minimum dari matriks asli, yaitu 100).
    - VT adalah transpos dari vektor singular kanan, berukuran 100×4600.

```python
# Membuat matriks Sigma dengan dimensi yang sesuai dengan U dan VT
Sigma = zeros((U.shape[1], VT.shape[0]))
```

- Sigma = zeros((U.shape[1], VT.shape[0])) membuat matriks Sigma dengan ukuran yang sama dengan U dan VT jika digabungkan, yaitu 100×4600.
- Sigma adalah matriks yang sebagian besar berisi nol, dan kita akan mengisi elemen-elemen diagonalnya dengan nilai singular dari s.

```python
# Mengisi Sigma dengan nilai-nilai singular
Sigma[:len(s), :len(s)] = diag(s)
```

- diag(s) membuat matriks diagonal dengan elemen diagonal yang diambil dari vektor s.
- Sigma[:len(s), :len(s)] = diag(s) menempatkan nilai-nilai singular dari s di sepanjang diagonal utama Sigma, sehingga Sigma sekarang berukuran 100×4600 dan mengandung semua nilai singular pada posisinya.

```python
# Merekonstruksi matriks
B = U.dot(Sigma.dot(VT))
```

- Sigma.dot(VT) mengalikan Sigma dengan VT, menghasilkan matriks perantara berukuran 100×4600.
- U.dot(...) kemudian mengalikan hasil tersebut dengan U, merekonstruksi matriks B yang sama ukurannya dengan matriks awal, yaitu 100×4600.
- B seharusnya mendekati matriks asli dfTFIDF karena proses SVD dan rekonstruksi matriks ini menghasilkan perkiraan dari matriks asli berdasarkan dekomposisi nilainya.
